```matlab
load NaresHighRes
islands = R(1:3);
land = union(R(4),R(5))
x = [-3 -3 3 3]*1e4;
y = [-1 1 1 -1]*5*1e4;
c2_boundary = [x; y];
boundary_poly = polyshape(c2_boundary');

target_concentration = 1;
[Ny, Nx] = size(target_concentration);
c = flipud(target_concentration);
x = min(c2_boundary(1,:)):(max(c2_boundary(1,:))-min(c2_boundary(1,:)))/Nx:max(c2_boundary(1,:));
y = min(c2_boundary(2,:)):(max(c2_boundary(2,:))-min(c2_boundary(2,:)))/Ny:max(c2_boundary(2,:));
c2_boundary_poly = polyshape(c2_boundary');
dx = x(2)-x(1);
dy = y(2)-y(1);

N = 10; ii = 1; jj = 1;
X = 0.975*dx/2*(2*rand(N,1)-1)+(x(ii)+x(ii+1))/2;
Y = 0.975*dy/2*(2*rand(N,1)-1)+(y(jj)+y(jj+1))/2;
in = inpolygon(X,Y,boundary_poly.Vertices(:,1),boundary_poly.Vertices(:,2));
X = X(in); Y = Y(in);
[~, b,~,~,~] = polybnd_voronoi([X Y],boundary_poly.Vertices);
count = 1; clear borders;
for m = 1:length(b)
    poly = polyshape(b{m});
    poly = intersect(poly,land);
    if area(poly)>0
        if length(regions(poly))>1
            R2 = regions(poly);
            for jj = 1:length(R2)
                borders(count) = R2(jj);
                count = count + 1;
            end
        else
            borders(count) = poly;
            count = count + 1;
        end
    end
end
```
